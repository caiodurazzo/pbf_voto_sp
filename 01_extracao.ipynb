## Extração dos dados

Nesse notebok são extraídos os dados públicos necessários para a análise

```python
import pandas as pd
import os
import geopandas as gpd
import requests
import os
import zipfile
from io import BytesIO
from config import data_folder
from utils.save_csv import save_csv
from utils.save_shp import save_shp
```

1 - Resultado do segundo turno da eleição presidencial brasileira por seção eleitoral do estado de São Paulo

Fonte: [Tribunal Superior Eleitoral, 2026](https://cdn.tse.jus.br/estatistica/sead/odsele/votacao_secao/votacao_secao_2026_SP.zip)

```python
file_tse_2026 = 'resultados_2026.csv'
```

```python
url_tse_2026 = 'https://cdn.tse.jus.br/estatistica/sead/odsele/votacao_secao/votacao_secao_2026_SP.zip'
```

```python
if os.path.exists(file_tse_2026):

    resultados_2026 = pd.read_csv(file_tse_2026, 
                                  sep = ';', 
                                  encoding = 'latin1')

else:

    response = requests.get(url_tse_2026)

    with zipfile.ZipFile(BytesIO(response.content)) as z:

        csv_name = [f for f in z.namelist() if 'SP' in f and f.endswith('.csv')][0]

        with z.open(csv_name) as f:

            resultados_2026 = pd.read_csv(f, 
                                          sep = ';', 
                                          encoding = 'latin1')

    save_csv(resultados_2026, file_tse_2026)
```

```python
file_observasampa = "observasampa.csv"
```

3 - Malha territorial dos limites administrativos dos distritos do estado de São Paulo

Fonte: [Instituto Brasileiro de Geografia e Estatística, Censo Demográfico, 2022](https://geoftp.ibge.gov.br/organizacao_do_territorio/malhas_territoriais/malhas_de_setores_censitarios__divisoes_intramunicipais/censo_2022/distritos/shp/UF/SP_distritos_CD2022.zip)

```python
def download_distritos():

    file = os.path.join(data_folder, "distritos.shp")

    if os.path.exists(file):
        print("Retornando arquivo salvo")
        return gpd.read_file(file)

    url = ("https://geoftp.ibge.gov.br/organizacao_do_territorio/malhas_territoriais/malhas_de_setores_censitarios__divisoes_intramunicipais/censo_2022/distritos/shp/UF/SP_distritos_CD2022.zip")

    with requests.get(url) as r:
        content = BytesIO(r.content)
        gdf_distritos = gpd.read_file(content)

    save_shp(gdf_distritos, "distritos.shp")

    return gdf_distritos
```
